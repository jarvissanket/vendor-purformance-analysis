###Understanding the Tables in the Dataset

#Purchases table
#The purchases table contains actual purchase data, including the date of purchase, products (brands) purchased by vendors, the amount paid (in dollars), and the quantity purchased.

#Purchase Prices table
#The purchase price column is derived from the purchase_prices table, which provides product-wise actual purchase prices. The combination of vendor and brand is unique in this table.

#Vendor Invoice table
#The vendor invoice table aggregates data from the purchases table, summarizing quantity and dollar amounts, along with an additional column for freight. This table maintains uniqueness based on vendor and PO (Purchase Order) number.

#Sales table
#The sales table captures actual sales transactions, detailing the brands sold, the quantity sold, the selling price, and the revenue earned.

```python
sales = data["sales"].sample(200_000, random_state=42)
purchases = data["purchases"].sample(200_000, random_state=42)
purchase_prices = data["purchase_prices"]
vendor_invoice = data["vendor_invoice"]
sales_agg = sales.groupby(
    ["VendorNo", "Brand"], as_index=False
).agg(
    TotalSalesQuantity=("SalesQuantity", "sum"),
    TotalSalesDollars=("SalesDollars", "sum"),
    TotalSalesPrice=("SalesPrice", "sum"),
    TotalExciseTax=("ExciseTax", "sum")
)

purchase_agg = purchases.groupby(
    ["VendorNumber", "Brand"], as_index=False
).agg(
    TotalPurchaseQuantity=("Quantity", "sum"),
    TotalPurchaseDollars=("Dollars", "sum")
)

freight_agg = vendor_invoice.groupby(
    "VendorNumber", as_index=False
).agg(
    TotalFreightCost=("Freight", "sum")
)
final_table = (
    purchase_prices
    .merge(
        sales_agg,
        left_on=["VendorNumber", "Brand"],
        right_on=["VendorNo", "Brand"],
        how="inner"
    )
    .merge(
        purchase_agg,
        on=["VendorNumber", "Brand"],
        how="inner"
    )
    .merge(
        freight_agg,
        on="VendorNumber",
        how="inner"
    )
)

final_table.head()

```

```python
# base tables
sales = data["sales"]
purchases = data["purchases"]
purchase_prices = data["purchase_prices"]
vendor_invoice = data["vendor_invoice"]

# 1. Aggregate sales
sales_agg = (
    sales
    .groupby(["VendorNo", "Brand"], as_index=False)
    .agg(
        TotalSalesQuantity=("SalesQuantity", "sum"),
        TotalSalesDollars=("SalesDollars", "sum"),
        TotalSalesPrice=("SalesPrice", "sum"),
        TotalExciseTax=("ExciseTax", "sum")
    )
)

# 2. Aggregate purchases
purchase_agg = (
    purchases
    .groupby(["VendorNumber", "Brand"], as_index=False)
    .agg(
        TotalPurchaseQuantity=("Quantity", "sum"),
        TotalPurchaseDollars=("Dollars", "sum")
    )
)

# 3. Aggregate freight
freight_agg = (
    vendor_invoice
    .groupby("VendorNumber", as_index=False)
    .agg(
        FreightCost=("Freight", "sum")
    )
)

# 4. Merge everything
vendor_sales_summary = (
    purchase_prices
    .merge(
        sales_agg,
        left_on=["VendorNumber", "Brand"],
        right_on=["VendorNo", "Brand"],
        how="left"
    )
    .merge(
        purchase_agg,
        on=["VendorNumber", "Brand"],
        how="left"
    )
    .merge(
        freight_agg,
        on="VendorNumber",
        how="left"
    )
)

vendor_sales_summary.head()

```

```python
purchase_merged = (
    data["purchases"]
    .merge(
        data["purchase_prices"],
        on=["VendorNumber", "Brand"],
        how="left",
        suffixes=("", "_pp")
    )
)
purchase_summary = (
    purchase_merged
    .groupby(
        [
            "VendorNumber",
            "VendorName",
            "Brand",
            "Description",
            "PurchasePrice",
            "Price",
            "Volume"
        ],
        as_index=False
    )
    .agg(
        TotalPurchaseQuantity=("Quantity", "sum"),
        TotalPurchaseDollars=("Dollars", "sum")
    )
    .rename(columns={"Price": "ActualPrice"})
)
sales_summary = (
    data["sales"]
    .groupby(["VendorNo", "Brand"], as_index=False)
    .agg(
        TotalSalesQuantity=("SalesQuantity", "sum"),
        TotalSalesDollars=("SalesDollars", "sum"),
        TotalSalesPrice=("SalesPrice", "sum"),
        TotalExciseTax=("ExciseTax", "sum")
    )
)
freight_summary = (
    data["vendor_invoice"]
    .groupby("VendorNumber", as_index=False)
    .agg(FreightCost=("Freight", "sum"))
)
vendor_sales_summary = (
    purchase_summary
    .merge(
        sales_summary,
        left_on=["VendorNumber", "Brand"],
        right_on=["VendorNo", "Brand"],
        how="left"
    )
    .merge(
        freight_summary,
        on="VendorNumber",
        how="left"
    )
    .sort_values(by="TotalPurchaseDollars", ascending=False)
)
vendor_sales_summary.head()
```

```python
def format_dollars(value):
    if value >=1_000_000:
        return f"{value / 1_000_000:.2f}M"
    elif value >=1_000:
        return f"{value / 1_000:.2f}K"
    else:
        return str(value)
```
